```python
from pathlib import Path
import pandas as pd
```

```python
OUTPUT_ROOT = Path("../data/outputs")

files = list(OUTPUT_ROOT.rglob("egg_prediction.csv"))

print(f"Files found: {len(files)}")
```

```python
dfs = []

for file in files:
    df_temp = pd.read_csv(file)
    dfs.append(df_temp)

df = pd.concat(dfs, ignore_index=True)

print(f"Total rows: {len(df)}")
df.head()
```

```python
df["inference_status"].value_counts(dropna=False)
```

```python
df["inference_status"].value_counts(
    dropna=False,
    normalize=True
) * 100
```

```python
pd.crosstab(
    df["farm_id"],
    df["inference_status"],
    margins=True
)
```
