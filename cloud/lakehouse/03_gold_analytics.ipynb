```python
# ============================================================
# CONFIGURATION
# ============================================================

# Define the ADLS paths used for reading Silver data
# and storing Gold analytical datasets.
account = "mastermmf001sta"

lakehouse_path = (
    f"abfss://lakehouse@{account}.dfs.core.windows.net"
)

silver_path = (
    f"{lakehouse_path}/silver/egg_predictions"
)

gold_daily_path = (
    f"{lakehouse_path}/gold/daily_production"
)

gold_spatial_path = (
    f"{lakehouse_path}/gold/spatial_monitoring"
)

spark.conf.set(
  "fs.azure.account.key.mastermmf001sta.dfs.core.windows.net",
  "Mkp5+QxUJpJ5/r64MPHfZruiqhiXr+jGHqV3yfUynm2pZG7KiY8t3k6lp9JHfI3YB78B4pPNr1er+AStK6ZnnA=="
)
```

```python
# ============================================================
# READ SILVER DATASET
# ============================================================

# Load the validated and enriched inference dataset
# generated in the Silver layer.
df_silver = (
    spark.read
    .format("delta")
    .load(silver_path)
)

print(f"Silver rows: {df_silver.count()}")

display(df_silver.limit(10))
```

```python
# ============================================================
# BUILD DAILY PRODUCTION GOLD DATASET
# ============================================================

# Aggregate Silver inference records by farm, house, battery
# and capture date to generate daily production and
# pipeline-quality metrics.
from pyspark.sql import functions as F

df_gold_daily = (
    df_silver
    .groupBy(
        "farm_id",
        "location",
        "house_number",
        "battery_number",
        "capture_date"
    )
    .agg(
        # Total number of inference records.
        F.count("*").alias("number_of_records"),

        # Number of successful and failed inferences.
        F.sum(
            F.when(F.col("inference_status") == "SUCCESS", 1).otherwise(0)
        ).alias("successful_inferences"),

        F.sum(
            F.when(F.col("inference_status") == "ERROR", 1).otherwise(0)
        ).alias("inference_errors"),

        # Mean egg count considering only available predictions.
        F.avg("egg_count").alias("mean_egg_count"),

        # Number of successful predictions with zero detected eggs.
        F.sum(
            F.when(
                (F.col("inference_status") == "SUCCESS")
                & (F.col("egg_count") == 0),
                1
            ).otherwise(0)
        ).alias("zero_egg_records")
    )
)
```

```python
# ============================================================
# CALCULATE DAILY QUALITY AND PRODUCTION METRICS
# ============================================================

# Calculate the percentage of inference errors and the percentage
# of successful predictions in which no eggs were detected.
df_gold_daily = (
    df_gold_daily

    .withColumn(
        "error_rate_pct",
        F.round(
            F.col("inference_errors")
            / F.col("number_of_records")
            * 100,
            2
        )
    )

    .withColumn(
        "zero_egg_rate_pct",
        F.round(
            F.col("zero_egg_records")
            / F.col("successful_inferences")
            * 100,
            2
        )
    )

    .withColumn(
        "mean_egg_count",
        F.round("mean_egg_count", 3)
    )
)
```

```python
# ============================================================
# REVIEW DAILY PRODUCTION METRICS
# ============================================================

# Display the Gold dataset ordered by farm, house,
# battery and date for an initial validation.
display(
    df_gold_daily
    .orderBy(
        "farm_id",
        "house_number",
        "battery_number",
        "capture_date"
    )
)
```

```python
# ============================================================
# VALIDATE DAILY GOLD DATASET
# ============================================================

# Check the number of aggregated records generated in Gold.
# Each row represents one farm, house, battery and capture date.
print(f"Gold daily rows: {df_gold_daily.count()}")

# Review the main analytical metrics to detect unexpected values.
display(
    df_gold_daily.select(
        "mean_egg_count",
        "error_rate_pct",
        "zero_egg_rate_pct"
    ).summary()
)
```

```python
# ============================================================
# REVIEW DAILY PRODUCTION BY FARM
# ============================================================

# Aggregate the successful inference results at farm and date level
# to inspect temporal changes in egg production.
df_farm_daily = (
    df_silver
    .filter(F.col("inference_status") == "SUCCESS")
    .groupBy(
        "farm_id",
        "capture_date"
    )
    .agg(
        F.round(
            F.avg("egg_count"),
            3
        ).alias("mean_egg_count")
    )
    .orderBy(
        "farm_id",
        "capture_date"
    )
)

display(df_farm_daily)
```

```python
# ============================================================
# WRITE DAILY PRODUCTION GOLD DATASET
# ============================================================

# Store the daily analytical metrics in the Gold layer
# using Delta Lake format.
(
    df_gold_daily
    .write
    .format("delta")
    .mode("overwrite")
    .save(gold_daily_path)
)

print(f"Gold daily rows stored: {df_gold_daily.count()}")
```

```python
# ============================================================
# VALIDATE DAILY PRODUCTION GOLD DATASET
# ============================================================

# Read the stored Delta dataset to verify that the Gold
# daily production table was created successfully.
df_gold_daily_check = (
    spark.read
    .format("delta")
    .load(gold_daily_path)
)

print(f"Gold daily rows stored: {df_gold_daily_check.count()}")

display(
    df_gold_daily_check
    .orderBy(
        "farm_id",
        "house_number",
        "battery_number",
        "capture_date"
    )
)
```

```python
# ============================================================
# CREATE SPATIAL POSITION GROUPS
# ============================================================

# Divide each battery side into groups of 50 consecutive positions.
# This provides enough spatial resolution to identify local patterns
# without keeping individual cage-level records in Gold.
df_spatial_base = (
    df_silver
    .withColumn(
        "position_group",
        (
            F.floor((F.col("position") - 1) / 50) * 50 + 1
        ).cast("int")
    )
)

display(
    df_spatial_base
    .select(
        "battery_number",
        "level",
        "side",
        "position",
        "position_group"
    )
    .orderBy("battery_number", "level", "side", "position")
    .limit(100)
)
```

```python
# ============================================================
# BUILD SPATIAL MONITORING GOLD DATASET
# ============================================================

# Aggregate inference results by physical area of the battery.
# Each row represents one level, side and position group
# for a specific farm, house, battery and date.
df_gold_spatial = (
    df_spatial_base
    .groupBy(
        "farm_id",
        "location",
        "house_number",
        "battery_number",
        "level",
        "side",
        "position_group",
        "capture_date"
    )
    .agg(
        F.count("*").alias("number_of_records"),

        F.sum(
            F.when(
                F.col("inference_status") == "SUCCESS", 1
            ).otherwise(0)
        ).alias("successful_inferences"),

        F.sum(
            F.when(
                F.col("inference_status") == "ERROR", 1
            ).otherwise(0)
        ).alias("inference_errors"),

        F.round(
            F.avg("egg_count"),
            3
        ).alias("mean_egg_count"),

        F.sum(
            F.when(
                (F.col("inference_status") == "SUCCESS")
                & (F.col("egg_count") == 0),
                1
            ).otherwise(0)
        ).alias("zero_egg_records")
    )
)
```

```python
# ============================================================
# CALCULATE SPATIAL MONITORING METRICS
# ============================================================

# Calculate inference error rate and the percentage of successful
# predictions with zero detected eggs for each physical area.
df_gold_spatial = (
    df_gold_spatial

    .withColumn(
        "error_rate_pct",
        F.round(
            F.col("inference_errors")
            / F.col("number_of_records")
            * 100,
            2
        )
    )

    .withColumn(
        "zero_egg_rate_pct",
        F.round(
            F.col("zero_egg_records")
            / F.col("successful_inferences")
            * 100,
            2
        )
    )
)
```

```python
# ============================================================
# REVIEW SPATIAL MONITORING DATASET
# ============================================================

# Review the resulting spatial dataset before storing it in Gold.
print(f"Gold spatial rows: {df_gold_spatial.count()}")

display(
    df_gold_spatial
    .orderBy(
        "farm_id",
        "house_number",
        "battery_number",
        "level",
        "side",
        "position_group",
        "capture_date"
    )
)
```

```python
# ============================================================
# VALIDATE SPATIAL GOLD DATASET
# ============================================================

# Review the range of the main spatial monitoring metrics
# before writing the dataset to the Gold layer.
display(
    df_gold_spatial.select(
        "mean_egg_count",
        "error_rate_pct",
        "zero_egg_rate_pct"
    ).summary()
)
```

```python
# ============================================================
# WRITE SPATIAL MONITORING GOLD DATASET
# ============================================================

# Store the spatial monitoring dataset in the Gold layer
# using Delta Lake format.
(
    df_gold_spatial
    .write
    .format("delta")
    .mode("overwrite")
    .save(gold_spatial_path)
)

print(f"Gold spatial rows stored: {df_gold_spatial.count()}")
```

```python
# ============================================================
# VALIDATE STORED SPATIAL GOLD DATASET
# ============================================================

# Read the stored Delta dataset to verify that the spatial
# monitoring table was created successfully.
df_gold_spatial_check = (
    spark.read
    .format("delta")
    .load(gold_spatial_path)
)

print(f"Gold spatial rows stored: {df_gold_spatial_check.count()}")

display(
    df_gold_spatial_check
    .orderBy(
        "farm_id",
        "house_number",
        "battery_number",
        "level",
        "side",
        "position_group",
        "capture_date"
    )
)
```
