```python
# ============================================================
# READ DATA FROM BRONZE
# ============================================================

df_bronze = (
    spark.read
    .format("delta")
    .load(bronze_path)
)

print(f"Bronze rows: {df_bronze.count()}")

display(df_bronze.limit(10))
```

```python
# ============================================================
# READ MASTER DATA FROM AZURE POSTGRESQL
# ============================================================

# JDBC connection parameters for the PostgreSQL database.
# Credentials should not be stored directly in the repository.
postgres_host = "mastermmf01.postgres.database.azure.com"
postgres_port = "5432"
postgres_database = "poultry_master_data"

jdbc_url = (
    f"jdbc:postgresql://{postgres_host}:{postgres_port}/{postgres_database}"
)

postgres_user = "monica"
postgres_password = "TFMtfm123"
```

```python
# ============================================================
# LOAD MASTER DATA QUERY
# ============================================================

# Retrieve the physical context of each cage by joining
# farm, house, battery and cage tables in PostgreSQL.
master_data_query = """
(
    SELECT
        f.name AS farm_id,
        h.house_number,
        c.cage_id,
        b.battery_number,
        c.level,
        c.side,
        c.position,
        f.location
    FROM cage c
    JOIN battery b
        ON c.battery_id = b.id
    JOIN house h
        ON b.house_id = h.id
    JOIN farm f
        ON h.farm_id = f.id
) AS master_data
"""
```

```python
# ============================================================
# READ MASTER DATA INTO SPARK
# ============================================================

# Read PostgreSQL master data into a Spark DataFrame using JDBC.
df_master = (
    spark.read
    .format("jdbc")
    .option("url", jdbc_url)
    .option("dbtable", master_data_query)
    .option("user", postgres_user)
    .option("password", postgres_password)
    .option("driver", "org.postgresql.Driver")
    .load()
)

print(f"Master data rows: {df_master.count()}")

display(df_master.limit(20))
```

```python
# ============================================================
# VALIDATE BRONZE AGAINST MASTER DATA
# ============================================================

# Extract the unique cage identifiers present in Bronze.
# A cage is functionally identified by farm_id, house_number and cage_id.
bronze_cages = (
    df_bronze
    .select(
        "farm_id",
        "house_number",
        "cage_id"
    )
    .distinct()
)

# Identify cages present in the inference data that do not have
# a corresponding record in the master data database.
unmatched_cages = (
    bronze_cages
    .join(
        df_master.select(
            "farm_id",
            "house_number",
            "cage_id"
        ),
        on=["farm_id", "house_number", "cage_id"],
        how="left_anti"
    )
)

print(f"Unique cages in Bronze: {bronze_cages.count()}")
print(f"Cages without master data: {unmatched_cages.count()}")

display(unmatched_cages)
```

```python
# ============================================================
# ENRICH BRONZE DATA WITH MASTER DATA
# ============================================================

# Join inference data from Bronze with the centralized master data.
# The functional key is composed of:
# farm_id + house_number + cage_id.
df_silver = (
    df_bronze
    .join(
        df_master,
        on=["farm_id", "house_number", "cage_id"],
        how="left"
    )
)

print(f"Silver rows after join: {df_silver.count()}")

display(df_silver.limit(20))
```

```python
# ============================================================
# VALIDATE MASTER DATA ENRICHMENT
# ============================================================

# Check that the physical context obtained from PostgreSQL
# is available for every inference record.
missing_master_context = (
    df_silver
    .filter(
        """
        battery_number IS NULL
        OR level IS NULL
        OR side IS NULL
        OR position IS NULL
        """
    )
)

print(
    f"Rows without complete master data context: "
    f"{missing_master_context.count()}"
)

display(missing_master_context)
```

```python
# ============================================================
# VALIDATE ROW COUNT AFTER JOIN
# ============================================================

# The enrichment join must not create or remove inference records.
bronze_count = df_bronze.count()
silver_count = df_silver.count()

print(f"Bronze rows: {bronze_count}")
print(f"Silver rows: {silver_count}")
print(f"Difference:  {silver_count - bronze_count}")
```

```python
# ============================================================
# VALIDATE INFERENCE STATUS
# ============================================================

# Count inference records by status to evaluate the quality
# of the predictions received from the Edge devices.
inference_status_summary = (
    df_silver
    .groupBy("inference_status")
    .count()
    .orderBy("inference_status")
)

display(inference_status_summary)
```

```python
# ============================================================
# CHECK NULL VALUES IN PREDICTION FIELDS
# ============================================================

# Prediction fields can be null when the inference process fails.
# These records are retained because they are useful for monitoring
# technical errors in the processing pipeline.
null_prediction_summary = (
    df_silver
    .selectExpr(
        "COUNT(*) AS total_rows",
        "SUM(CASE WHEN egg_count IS NULL THEN 1 ELSE 0 END) AS null_egg_count",
        "SUM(CASE WHEN confidence IS NULL THEN 1 ELSE 0 END) AS null_confidence",
        "SUM(CASE WHEN error_message IS NOT NULL THEN 1 ELSE 0 END) AS rows_with_error"
    )
)

display(null_prediction_summary)
```

```python
# ============================================================
# CHECK DUPLICATE INFERENCE RECORDS
# ============================================================

# Each cage should have only one inference result for a given
# capture date and image.
duplicate_records = (
    df_silver
    .groupBy(
        "farm_id",
        "house_number",
        "cage_id",
        "capture_date",
        "image_name"
    )
    .count()
    .filter("count > 1")
)

print(f"Duplicate inference records: {duplicate_records.count()}")

display(duplicate_records)
```

```python
# ============================================================
# VALIDATE CONSISTENCY OF INFERENCE RESULTS
# ============================================================

# Successful inferences should contain an egg count.
# Failed inferences should contain an error message and no egg count.
inference_consistency = (
    df_silver
    .selectExpr(
        """
        SUM(
            CASE
                WHEN inference_status = 'SUCCESS'
                AND egg_count IS NULL
                THEN 1
                ELSE 0
            END
        ) AS success_without_egg_count
        """,
        """
        SUM(
            CASE
                WHEN inference_status = 'ERROR'
                AND error_message IS NULL
                THEN 1
                ELSE 0
            END
        ) AS error_without_message
        """,
        """
        SUM(
            CASE
                WHEN inference_status = 'ERROR'
                AND egg_count IS NOT NULL
                THEN 1
                ELSE 0
            END
        ) AS error_with_egg_count
        """
    )
)

display(inference_consistency)
```

```python
# ============================================================
# WRITE SILVER DATASET TO DELTA LAKE
# ============================================================

# Store the validated and enriched inference dataset in the
# Silver layer using Delta Lake format.
(
    df_silver
    .write
    .format("delta")
    .mode("overwrite")
    .save(silver_path)
)
```

```python
# ============================================================
# VALIDATE SILVER DATASET
# ============================================================

# Read the stored Silver dataset again to verify that
# the Delta table was created successfully.
df_silver_check = (
    spark.read
    .format("delta")
    .load(silver_path)
)

print(f"Silver rows stored: {df_silver_check.count()}")

display(df_silver_check.limit(20))
```

```python
# ============================================================
# SILVER PROCESSING SUMMARY
# ============================================================

# Summarize the main validation results obtained during
# the construction of the Silver layer.
silver_summary = [
    ("Bronze records", df_bronze.count()),
    ("Silver records", df_silver_check.count()),
    ("Unmatched cages", unmatched_cages.count()),
    ("Duplicate records", duplicate_records.count()),
    ("Inference errors", df_silver.filter("inference_status = 'ERROR'").count()),
]

display(
    spark.createDataFrame(
        silver_summary,
        ["Validation", "Count"]
    )
)
```
