```python
# ============================================================
# INPUT FILE PATTERN
# ============================================================

# Edge output structure:
# <farm>/<year>/<month>/<day>/egg_prediction.csv

input_path = (
    f"{landing_path}/*/*/*/*/egg_prediction.csv"
)

print(f"Input pattern: {input_path}")
```

```python
# ============================================================
# READ LANDING DATA
# ============================================================

# Read all Edge inference CSV files available in the landing zone.
# Schema inference is enabled for this prototype.

df_landing = (
    spark.read
    .option("header", True)
    .option("inferSchema", True)
    .csv(input_path)
)

display(df_landing)
```

```python
# ============================================================
# BASIC INGESTION CHECKS
# ============================================================

total_rows = df_landing.count()

print(f"Total rows loaded: {total_rows}")

print("\nDetected farms:")
df_landing.select("farm_id").distinct().show()

print("\nDetected dates:")
df_landing.select("capture_date").distinct().orderBy(
    "capture_date"
).show()
```

```python
# ============================================================
# CHECK INPUT SCHEMA
# ============================================================

df_landing.printSchema()
```

```python
# ============================================================
# ADD INGESTION METADATA
# ============================================================

from pyspark.sql.functions import current_timestamp, input_file_name


df_bronze = (
    df_landing
    .withColumn(
        "source_file",
        input_file_name(),
    )
    .withColumn(
        "ingestion_timestamp",
        current_timestamp(),
    )
)

display(df_bronze)
```

```python
# ============================================================
# BRONZE VALIDATION
# ============================================================

print(f"Landing rows: {df_landing.count()}")
print(f"Bronze rows:  {df_bronze.count()}")

if df_landing.count() != df_bronze.count():
    raise RuntimeError(
        "Row count changed during Bronze ingestion."
    )

print("Bronze ingestion validation completed successfully.")
```

```python
# ============================================================
# BRONZE CONFIGURATION
# ============================================================

# ADLS container used for the Medallion architecture
lakehouse_path = (
    f"abfss://lakehouse@{account}.dfs.core.windows.net"
)

# Bronze layer for Edge inference results
bronze_path = (
    f"{lakehouse_path}/bronze/egg_predictions"
)

print(f"Bronze path: {bronze_path}")
```

```python
# ============================================================
# WRITE BRONZE DATA
# ============================================================

# Store the ingested Edge data in Delta format.
# Overwrite mode is used for the prototype execution.

(
    df_bronze
    .write
    .format("delta")
    .mode("overwrite")
    .save(bronze_path)
)

print("Bronze data written successfully.")
```

```python
# ============================================================
# VERIFY BRONZE DATA
# ============================================================

# Read the Delta dataset again to verify that it was
# successfully stored in the Bronze layer.

df_bronze_check = (
    spark.read
    .format("delta")
    .load(bronze_path)
)

print(f"Bronze rows stored: {df_bronze_check.count()}")

display(df_bronze_check.limit(20))
```
